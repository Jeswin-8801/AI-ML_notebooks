```python
import matplotlib.pyplot as plt
import numpy as np

# Data Setup
v = np.array([[1, 1, 0], [1, 0, 1], [0, 1, 1]]) # Original Basis V
e1 = v[0] / np.linalg.norm(v[0])
e2 = np.array([0.5, -0.5, 1]) / np.linalg.norm([0.5, -0.5, 1])
e3 = np.array([-2/3, 2/3, 2/3]) / np.linalg.norm([-2/3, 2/3, 2/3])
e = np.array([e1, e2, e3]) # Orthonormal Basis E

colors = ['blue', 'red', 'green']

# Figure with Subplots
fig = plt.figure(figsize=(15, 10))
views = [
    {'title': 'Isometric View (3D)', 'elev': 30, 'azim': 45},
    {'title': 'Top View (XY Plane)', 'elev': 90, 'azim': 0},
    {'title': 'Side View (XZ Plane)', 'elev': 0, 'azim': 0},
    {'title': 'Front View (YZ Plane)', 'elev': 0, 'azim': 90}
]

for i, view in enumerate(views):
    ax = fig.add_subplot(2, 2, i+1, projection='3d')
    for j in range(3):
        # Plot Original (Dashed)
        ax.quiver(0, 0, 0, v[j,0], v[j,1], v[j,2], color=colors[j], linestyle='--', arrow_length_ratio=0.1)
        # Plot Orthonormal (Solid)
        ax.quiver(0, 0, 0, e[j,0], e[j,1], e[j,2], color=colors[j], linestyle='-', arrow_length_ratio=0.1)
    
    # Labeling and Formatting
    ax.set_title(view['title'])
    ax.view_init(elev=view['elev'], azim=view['azim'])
    ax.set_xlim([-0.5, 1.2]); ax.set_ylim([-0.5, 1.2]); ax.set_zlim([-0.5, 1.2])
    if view['elev'] != 90: ax.set_zlabel('Z')
    if view['azim'] != 90: ax.set_ylabel('Y')
    ax.set_xlabel('X')

plt.tight_layout()
plt.show()
```
